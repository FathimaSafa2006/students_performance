Student Performance Analysis Using Pandas, NumPy, and Matplotlib

Educational institutions generate large amounts of student-related data that can be analyzed to improve academic performance and learning outcomes. This project focuses on analyzing student performance factors using Python libraries such as Pandas, NumPy, and Matplotlib.

The dataset contains information related to:

Study hours

Attendance

Sleep hours

Motivation level

Teacher quality

Tutoring sessions

Internet access

Exam scores


The main objective of this project is to identify the major factors that influence student exam performance and visualize the relationships between different variables.

This project demonstrates:

Data loading and cleaning

Statistical analysis

Correlation analysis

Data visualization

Insight generation


The analysis helps in understanding how different academic and lifestyle factors affect student success.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("/content/drive/MyDrive/Data Analyst/Data Sets/project dataanalyst/StudentPerformanceFactors (1).csv")

print(df.head())
```

```python
print(df.shape)
```

```python
print(df.columns)
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
print("Average Score:", np.mean(df['Exam_Score']))
```

```python
print("Highest Score:", np.max(df['Exam_Score']))
```

```python
print("Lowest Score:", np.min(df['Exam_Score']))
```

```python
print("Standard Deviation:", np.std(df['Exam_Score']))
```

```python
attendance_corr = df['Attendance'].corr(df['Exam_Score'])

print(attendance_corr)
```

```python

plt.figure(figsize=(7,5))
plt.scatter(df['Attendance'], df['Exam_Score'])

plt.xlabel("Attendance")
plt.ylabel("Exam Score")
plt.title("Attendance vs Exam Score")

plt.show()
```

```python
study_corr = df['Hours_Studied'].corr(df['Exam_Score'])

print(study_corr)
```

```python

plt.figure(figsize=(7,5))
plt.scatter(df['Hours_Studied'], df['Exam_Score'])

plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.title("Study Hours vs Exam Score")

plt.show()
```

```python
gender_avg = df.groupby('Gender')['Exam_Score'].mean()

print(gender_avg)
```

```python

plt.figure(figsize=(6,4))
plt.bar(gender_avg.index, gender_avg.values)

plt.xlabel("Gender")
plt.ylabel("Average Exam Score")
plt.title("Gender vs Exam Score")

plt.show()
```

```python
sleep_avg = df.groupby('Sleep_Hours')['Exam_Score'].mean()

print(sleep_avg)
```

```python

plt.figure(figsize=(8,5))
plt.plot(sleep_avg.index, sleep_avg.values, marker='o')

plt.xlabel("Sleep Hours")
plt.ylabel("Average Exam Score")
plt.title("Sleep Hours vs Exam Score")

plt.show()
```

```python
motivation_avg = df.groupby('Motivation_Level')['Exam_Score'].mean()

print(motivation_avg)
```

```python

plt.figure(figsize=(6,4))
plt.bar(motivation_avg.index, motivation_avg.values)

plt.xlabel("Motivation Level")
plt.ylabel("Average Exam Score")
plt.title("Motivation Level vs Exam Score")

plt.show()
```

```python
teacher_avg = df.groupby('Teacher_Quality')['Exam_Score'].mean()

print(teacher_avg)
```

```python

plt.figure(figsize=(6,4))
plt.bar(teacher_avg.index, teacher_avg.values)

plt.xlabel("Teacher Quality")
plt.ylabel("Average Exam Score")
plt.title("Teacher Quality vs Exam Score")

plt.show()
```

```python
internet_avg = df.groupby('Internet_Access')['Exam_Score'].mean()

print(internet_avg)
```

```python

plt.figure(figsize=(5,4))
plt.bar(internet_avg.index, internet_avg.values)

plt.xlabel("Internet Access")
plt.ylabel("Average Exam Score")
plt.title("Internet Access vs Exam Score")

plt.show()
```

```python
plt.figure(figsize=(7,5))

plt.hist(df['Exam_Score'], bins=10)

plt.xlabel("Exam Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Exam Scores")

plt.show()
```
